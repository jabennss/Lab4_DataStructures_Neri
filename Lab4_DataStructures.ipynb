```python
LAST_NAME = "Neri"
SEED_NUM = 9
FAVORITE_ARTIST = "IV OF SPADES"

# Generate Test Sources
test_sources = [f"{LAST_NAME}_{label}" for label in ["T1", "T2", "T3", "T4", "T5"]]

# Measurement Categories
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

# Generate Measurement Values from character positions
values = []
for i in range(5):
    char = FAVORITE_ARTIST[i]
    val = SEED_NUM + (ord(char) % 20)
    values.append(round(val, 2))

# Build Records
test_records = []
for src in test_sources:
    for cat in categories[:3]:
        test_records.append({"source": src, "category": cat, "value": values[categories.index(cat)]})

# Add duplicate record
test_records.append({"source": "Neri_T1", "category": "TEMPERATURE", "value": values[0]})

# Add more categories for T1 and T2
for src in test_sources[:2]:
    for cat in categories[3:]:
        test_records.append({"source": src, "category": cat, "value": values[categories.index(cat)]})

# === DISPLAY RESULTS ===
print("="*60)
print("EXERCISE 1 — ENGINEERING TEST DATA ORGANIZER")
print("="*60)

# 1. Generated Test Data
print("\n[1] GENERATED TEST DATA")
for rec in test_records:
    print(f"  {rec}")

# 2. Organized Test Data
print("\n[2] ORGANIZED TEST DATA (by Source)")
organized = {}
for rec in test_records:
    organized.setdefault(rec["source"], []).append(rec)
for src, recs in organized.items():
    print(f"  {src}:")
    for r in recs:
        print(f"    → {r['category']}: {r['value']}")

# 3. Repeated / Distinct Data
print("\n[3] REPEATED & DISTINCT DATA")
from collections import Counter
record_tuples = [(r["source"], r["category"]) for r in test_records]
counts = Counter(record_tuples)
repeated = [k for k, v in counts.items() if v > 1]
distinct = len(set(record_tuples))
print(f"  Repeated Entries: {repeated if repeated else 'None'}")
print(f"  Distinct Combinations: {distinct}")

# 4. Numerical & Categorical Summaries
print("\n[4] NUMERICAL & CATEGORICAL SUMMARIES")
all_values = [r["value"] for r in test_records]
print(f"  Total Records: {len(test_records)}")
print(f"  Min Value: {min(all_values)}")
print(f"  Max Value: {max(all_values)}")
print(f"  Avg Value: {round(sum(all_values)/len(all_values), 2)}")
print(f"  Categories Used: {', '.join(sorted(set(r['category'] for r in test_records)))}")
print(f"  Test Sources: {', '.join(set(r['source'] for r in test_records))}")

# 5. Processed Results
print("\n[5] PROCESSED RESULTS (Sorted)")
sorted_records = sorted(test_records, key=lambda x: (x["source"], x["category"]))
for rec in sorted_records:
    print(f"  {rec['source']} | {rec['category']:12} | {rec['value']}")

# 6. Ordered Results — same as above ✅

# 7. Final Summary
print("\n[7] FINAL SUMMARY")
print(f"  Total Records Collected: {len(test_records)}")
print(f"  Unique Test Sources: {len(organized)}")
print(f"  Unique Categories: {len(set(r['category'] for r in test_records))}")
print(f"  Duplicate Records Found: {len(repeated)}")
print(f"  Value Range: {min(all_values)} to {max(all_values)}")
print("="*60)
```

```python
LAST_NAME = "Neri"
SEED_NUM = 9
FAVORITE_ARTIST = "IV OF SPADES"

# Student Names
students = [f"{LAST_NAME}_{label}" for label in ["S1", "S2", "S3", "S4", "S5"]]

# Subjects
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

# Generate Scores
performance_records = []
for i, student in enumerate(students):
    for j, subj in enumerate(subjects):
        char_index = (i + j) % len(FAVORITE_ARTIST)
        char = FAVORITE_ARTIST[char_index]
        score = SEED_NUM + (ord(char) % 31) + 60  # Base score in reasonable range
        performance_records.append({"student": student, "subject": subj, "score": score})

# Add repeated record
performance_records.append({"student": "Neri_S1", "subject": "PROGRAMMING", "score": 88})

# Add incomplete record
performance_records.append({"student": "Neri_S5", "subject": "ENGLISH", "score": None})

# Performance Category Function
def get_category(score):
    if score is None: return "INCOMPLETE"
    if score >= 90: return "EXCELLENT"
    if score >= 75: return "SATISFACTORY"
    return "NEEDS IMPROVEMENT"

# === DISPLAY RESULTS ===
print("="*60)
print("EXERCISE 2 — STUDENT PERFORMANCE DATA ANALYZER")
print("="*60)

# 1. Generated Performance Records
print("\n[1] GENERATED PERFORMANCE RECORDS")
for rec in performance_records:
    cat = get_category(rec["score"])
    print(f"  {rec['student']:12} | {rec['subject']:15} | Score: {rec['score'] or '—':>4} | {cat}")

# 2. Organized Student/Subject Data
print("\n[2] ORGANIZED DATA (by Student)")
by_student = {}
for rec in performance_records:
    by_student.setdefault(rec["student"], []).append(rec)
for student, recs in by_student.items():
    print(f"  {student}:")
    for r in recs:
        print(f"    → {r['subject']}: {r['score'] or 'MISSING'} ({get_category(r['score'])})")

# 3. Repeated & Incomplete Records
print("\n[3] REPEATED & INCOMPLETE RECORDS")
rec_tuples = [(r["student"], r["subject"]) for r in performance_records]
from collections import Counter
counts = Counter(rec_tuples)
repeated = [k for k, v in counts.items() if v > 1]
incomplete = [r for r in performance_records if r["score"] is None]
print(f"  Repeated Records: {repeated if repeated else 'None'}")
print(f"  Incomplete Records: {[(r['student'], r['subject']) for r in incomplete] if incomplete else 'None'}")

# 4. Performance Categories
print("\n[4] PERFORMANCE CATEGORIES SUMMARY")
cat_counts = {}
for rec in performance_records:
    c = get_category(rec["score"])
    cat_counts[c] = cat_counts.get(c, 0) + 1
for cat, cnt in cat_counts.items():
    print(f"  {cat}: {cnt} record(s)")

# 5. Student & Subject Summaries
print("\n[5] STUDENT SUMMARIES")
for student in students:
    scores = [r["score"] for r in by_student.get(student, []) if r["score"] is not None]
    if scores:
        print(f"  {student}: Avg = {round(sum(scores)/len(scores),2)} | Total = {len(scores)} subjects")
    else:
        print(f"  {student}: No complete records")

print("\n  SUBJECT SUMMARIES")
for subj in subjects:
    scores = [r["score"] for r in performance_records if r["subject"] == subj and r["score"] is not None]
    if scores:
        print(f"  {subj:15} | Avg: {round(sum(scores)/len(scores),2)} | Highest: {max(scores)}")

# 6. Ordered Results
print("\n[6] ORDERED RESULTS")
sorted_records = sorted(performance_records, key=lambda x: (x["student"], x["subject"]))
for rec in sorted_records:
    print(f"  {rec['student']:12} | {rec['subject']:15} | {rec['score'] or '—':>4} | {get_category(rec['score'])}")

# 7. Final Report
print("\n[7] FINAL STUDENT PERFORMANCE REPORT")
print(f"  Total Records: {len(performance_records)}")
print(f"  Complete Records: {len(performance_records) - len(incomplete)}")
print(f"  Missing Scores: {len(incomplete)}")
print(f"  Duplicate Entries: {len(repeated)}")
print(f"  Students Analyzed: {len(students)}")
print(f"  Subjects Covered: {len(subjects)}")
print("="*60)
```

```python
LAST_NAME = "Neri"
SEED_NUM = 9
FAVORITE_ARTIST = "IV OF SPADES"

# Customers
customers = [f"{LAST_NAME}_{label}" for label in ["C1", "C2", "C3", "C4", "C5"]]

# Products & Prices
products = {
    "NOTEBOOK": 45.00,
    "PEN": 12.50,
    "USB": 85.00,
    "HEADPHONES": 350.00,
    "CALCULATOR": 120.00
}
product_list = list(products.keys())

# Generate Quantities from character positions
quantities = []
for i in range(5):
    char = FAVORITE_ARTIST[i]
    qty = SEED_NUM + (ord(char) % 8)+ 1
    quantities.append(qty)

# Generate Transactions
import random
random.seed(SEED_NUM)
transactions = []
for cust in customers:
    for i in range(random.randint(1, 3)):
        prod = random.choice(product_list)
        qty = quantities[product_list.index(prod)]
        transactions.append({
            "customer": cust,
            "product": prod,
            "quantity": qty,
            "unit_price": products[prod],
            "total_value": round(qty * products[prod], 2)
        })

# Add Repeated Purchase
transactions.append({
    "customer": "Neri_C1",
    "product": "NOTEBOOK",
    "quantity": quantities[0],
    "unit_price": products["NOTEBOOK"],
    "total_value": round(quantities[0] * products["NOTEBOOK"], 2)
})

# === DISPLAY RESULTS ===
print("="*60)
print("EXERCISE 3 — TRANSACTION HISTORY ANALYZER")
print("="*60)

# 1. Generated Transaction Records
print("\n[1] GENERATED TRANSACTION RECORDS")
for i, tx in enumerate(transactions, 1):
    print(f"  {i:2}. {tx['customer']:10} | {tx['product']:12} | Qty: {tx['quantity']:2} | "
          f"₱{tx['unit_price']:>6.2f} | Total: ₱{tx['total_value']:>7.2f}")

# 2. Organized Transaction Data
print("\n[2] ORGANIZED DATA (by Customer)")
by_customer = {}
for tx in transactions:
    by_customer.setdefault(tx["customer"], []).append(tx)
for cust, txs in by_customer.items():
    print(f"  {cust}:")
    for t in txs:
        print(f"    → {t['product']} × {t['quantity']} = ₱{t['total_value']:.2f}")

# 3. Repeated & Distinct Transactions
print("\n[3] REPEATED & DISTINCT TRANSACTIONS")
tx_signatures = [(t["customer"], t["product"]) for t in transactions]
from collections import Counter
tx_counts = Counter(tx_signatures)
repeated = [k for k, v in tx_counts.items() if v > 1]
distinct_prods = set(t["product"] for t in transactions)
distinct_custs = set(t["customer"] for t in transactions)
print(f"  Repeated Purchases: {repeated if repeated else 'None'}")
print(f"  Distinct Products Sold: {len(distinct_prods)} — {', '.join(distinct_prods)}")
print(f"  Distinct Customers: {len(distinct_custs)} — {', '.join(distinct_custs)}")

# 4. Transaction Values
print("\n[4] TRANSACTION VALUES")
total_revenue = sum(t["total_value"] for t in transactions)
avg_transaction = total_revenue / len(transactions)
print(f"  Total Transactions: {len(transactions)}")
print(f"  Total Revenue: ₱{total_revenue:.2f}")
print(f"  Average Transaction: ₱{avg_transaction:.2f}")

# 5. Customer Summaries
print("\n[5] CUSTOMER SUMMARIES")
for cust in customers:
    txs = by_customer.get(cust, [])
    spent = sum(t["total_value"] for t in txs)
    items = sum(t["quantity"] for t in txs)
    print(f"  {cust:10} | Orders: {len(txs)} | Items: {items:2} | Spent: ₱{spent:.2f}")

# 6. Product Summaries
print("\n[6] PRODUCT SUMMARIES")
for prod in product_list:
    sold = sum(t["quantity"] for t in transactions if t["product"] == prod)
    revenue = sold * products[prod]
    print(f"  {prod:12} | Sold: {sold:2} | Revenue: ₱{revenue:.2f}")

# 7. Ordered Results
print("\n[7] ORDERED RESULTS (Highest Value First)")
sorted_tx = sorted(transactions, key=lambda x: x["total_value"], reverse=True)
for i, tx in enumerate(sorted_tx, 1):
    print(f"  {i:2}. {tx['customer']:10} | {tx['product']:12} | ₱{tx['total_value']:>7.2f}")

# 8. Overall Summary
print("\n[8] OVERALL TRANSACTION SUMMARY")
print(f"  Total Transactions: {len(transactions)}")
print(f"  Unique Customers: {len(distinct_custs)}")
print(f"  Unique Products: {len(distinct_prods)}")
print(f"  Total Revenue: ₱{total_revenue:.2f}")
top_product = max(product_list, key=lambda p: sum(t['quantity'] for t in transactions if t['product'] == p))
top_customer = max(customers, key=lambda c: sum(t['total_value'] for t in transactions if t['customer'] == c))
print(f"  Top Product: {top_product}")
print(f"  Top Customer: {top_customer}")
print("="*60)
```
